```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
DATA_PATH = "../data/MTA_Subway_Customer_Journey-Focused_Metrics__2020-2024.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
df["month"] = pd.to_datetime(df["month"])

df["year"] = df["month"].dt.year
df["month_num"] = df["month"].dt.month

df[["month", "year", "month_num"]].head()
```

```python
df["line"] = (
    df["line"]
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.upper()
)

shuttle_map = {
    "S 42ND": "S_42ND",
    "S FKLN": "S_FKLN",
    "S ROCK": "S_ROCK",
}

df["line_clean"] = df["line"].replace(shuttle_map)

sorted(df["line_clean"].unique())
```

```python
def line_type(x):
    if x.isdigit():
        return "NUMBER"
    if x.startswith("S"):
        return "SHUTTLE"
    if x == "JZ":
        return "COMBINED"
    return "LETTER"

df["line_type"] = df["line_clean"].apply(line_type)

df[["line_clean", "line_type"]].drop_duplicates().sort_values("line_clean")
```

```python
CLEAN_PATH = "../data/mta_cleaned.csv"
df.to_csv(CLEAN_PATH, index=False)
```
